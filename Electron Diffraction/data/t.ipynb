```typescript
class DataSheet {
    // constructor(columns: Number) {
    constructor(titles: String[]) { // e.g. ["Voltage (V)", "D1 (M)", "D2 (M)"]
        this.titles = titles;
        this.columns = titles.length;
        this.data = [];
    }

    get rows() {
        return this.data.length;
    }

    insert() {
        const data = [...arguments];

        if (data.length > this.columns) throw new Error('Entry exceeds allocated columns');
        if (data.length < this.columns) throw new Error(`Entry does not meet required columns (${this.columns})`);

        this.data.push(data);
    }

    get parsed() {
        return [this.titles].concat(this.data).map((row: any[]) => row.join('\t')).join('\n');
    }
}
```

```typescript
const avg = nums => nums.reduce((a, b) => a + b, 0) / nums.length;

function pxToM(PX, PX10MM) {
    const factor = 10 / PX10MM;

    return PX * factor * 1e-3;
}
```
